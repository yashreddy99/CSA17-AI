```
%%writefile hanoi.pl
hanoi(1, A, _, C) :-
    write('Move disk from '), write(A),
    write(' to '), write(C), nl.

hanoi(N, A, B, C) :-
    N > 1,
    M is N - 1,
    hanoi(M, A, C, B),
    write('Move disk from '), write(A),
    write(' to '), write(C), nl,
    hanoi(M, B, A, C).
```

```
import subprocess

# Using subprocess to capture the output from swipl
try:
    cmd = f"swipl -s hanoi.pl -g \"hanoi({n}, 'A', 'B', 'C').\" -t halt"
    result = subprocess.check_output(cmd, shell=True, stderr=subprocess.STDOUT, text=True)
    print(result)
except subprocess.CalledProcessError as e:
    print(f"Error executing Prolog: {e.output}")
```
